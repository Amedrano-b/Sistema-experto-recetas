```python
import pandas as pd
import ast

#Leer la base de datos
file_path = "DBrecipes1.csv"
recipe_df = pd.read_csv(file_path, encoding='latin-1')

recipe_df.head()
```

```python
recipe_df["ingredients_quantities"] = recipe_df["ingredients_quantities"].apply(ast.literal_eval)
recipe_df["ingredients_list"] = recipe_df["ingredients_list"].apply(ast.literal_eval)

def combine_ingredients(row):
    quantities = row["ingredients_quantities"]
    ingredients = row["ingredients_list"]
    combined = []

    # Une cantidad con ingrediente (maneja si tienen longitudes diferentes)
    for i in range(min(len(quantities), len(ingredients))):
        combined.append(f"{quantities[i]} de {ingredients[i]}")

    return combined

recipe_df["ingredients_combined"] = recipe_df.apply(combine_ingredients, axis=1)
recipe_df["ingredients_text_combined"] = recipe_df["ingredients_combined"].apply(lambda x: ", ".join(x))

recipe_df["ingredients_text_combined"]
```

```python
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
recipe_df.dtypes
```

```python
 
print(str)
#Preprocess ingredents
recipe_df["ingredients_list"] = recipe_df["ingredients_list"].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else x
)

# Asegurar que no haya NaN ni floats
recipe_df["ingredients_list"] = recipe_df["ingredients_list"].apply(
    lambda x: x if isinstance(x, list) else []
)
recipe_df["text_only"] = recipe_df["ingredients_list"].apply(
    lambda lst: " ".join(lst).lower()
)

# Vectorizar
vectorizer = TfidfVectorizer()
x_ingredients = vectorizer.fit_transform(recipe_df["text_only"])

x_ingredients.shape
```

```python
scaler = StandardScaler()
x_numerical = scaler.fit_transform(recipe_df[["calories", "servings", "prep_time_hour", "prep_time_min"]])
```

```python
x_numerical
```

```python
x_combined = np.hstack([x_numerical, x_ingredients.toarray()])
x_combined
```

```python
#Train KNN model
knn = NearestNeighbors(n_neighbors=3, metric="euclidean")
knn.fit(x_combined)
```

```python
def recommend_recipes(input_features):
    input_features_scaled =scaler.transform([input_features[:4]])
    input_ingredients_transformed = vectorizer.transform([input_features[4]])

    input_combined = np.hstack([input_features_scaled, input_ingredients_transformed.toarray()])
    distances, indeces = knn.kneighbors(input_combined)

    recommendations =recipe_df.iloc[indeces[0]]

    return recommendations[["recipe_name", "calories", "servings", "prep_time_hour", "prep_time_min", "ingredients_text_combined", "image_url"]]

#Example Input
input_features = [350,6,0,50, 'leche, huevo, canela, azucar']
recommendations = recommend_recipes(input_features)
recommendations
```
